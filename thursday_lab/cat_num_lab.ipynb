```python
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

rng = np.random.default_rng(seed = 100)

from scipy.stats import beta
from scipy.stats import uniform
import math
```

# Q1

```python
K = 10
n = 5000
Q1list = np.arange(1, K + 1) / K
Q1picks = np.random.choice(Q1list, size = n)
sns.ecdfplot(Q1picks)
```

```python
K = 1000
n = 5000
Q1list = np.arange(1, K + 1) / K
Q1picks = np.random.choice(Q1list, size = n)
sns.ecdfplot(Q1picks)
```

As we increase K, this approaches the normal distribution.

# Q2

```python
n = 32
Q2 = rng.uniform(- math.sqrt(3), math.sqrt(3), size=(5000, n))
Q2processed = [Q.mean() * np.sqrt(n) for Q in Q2]
sns.ecdfplot(Q2processed)
```

Let's increase n:

```python
n = 1000
Q2 = rng.uniform(- math.sqrt(3), math.sqrt(3), size=(5000, n))
Q2processed = [Q.mean() * np.sqrt(n) for Q in Q2]

Q2processed = np.sqrt(n) * Q2.mean(axis=1)
sns.ecdfplot(Q2processed)
```

As n increases, this approaches the normal distribution.

# Q3

```python
HT = [1, 0]
n = 40
b = 5000
p = .5
probabilities = [p, 1 - p]
Q3 = rng.choice(HT, p=probabilities, size=(b, n))

def q3alt(arr):
    return (sum(arr) - (n * p))/(math.sqrt( n * p * (1-p) ))

newQ3 = [q3alt(Q) for Q in Q3]
sns.ecdfplot(newQ3)
```

Let's increase n:

```python
HT = [1, 0]
n = 400
b = 5000
p = .5
probabilities = [p, 1 - p]
Q3 = rng.choice(HT, p=probabilities, size=(b, n))

def q3alt(arr):
    return (sum(arr) - (n * p))/(math.sqrt( n * p * (1-p) ))

newQ3 = [q3alt(Q) for Q in Q3]
sns.ecdfplot(newQ3)
```

Let's change the probabilities to .9 and .1:

```python
HT = [1, 0]
n = 400
b = 5000
p = .9
probabilities = [p, 1 - p]
Q3 = rng.choice(HT, p=probabilities, size=(b, n))

def q3alt(arr):
    return (sum(arr) - (n * p))/(math.sqrt( n * p * (1-p) ))

newQ3 = [q3alt(Q) for Q in Q3]
sns.ecdfplot(newQ3)
```

This approaches a normal distribution as n increases. Changing p did not change the results of the ECDF.

# Q5

```python
k = 20 #20 storms in each year
n = 5000 
Q5 = rng.uniform(0, 1, size=(n, k))
Q5.sort(axis=1)
for i in range(k):
    sns.ecdfplot(Q5[:,i])
plt.show()
```

This process generates the beta distribution. The distribution is skewed closer to 1 for a larger rank and closer to 0 for a smaller rank.

# Q6

Since (e^delta)(e^delta)(e^-delta) = e^(delta + delta - delta), we can generate n step (in this case 1/delta) values of either 1 or -1 (with an equal 50/50 chance of being selected) and then sum all of these values. Once we have the sum, we can multiply it by the square root of delta before plugging that result in as the exponent for e. This will be done B = 1,000 times and then those 1,000 values will be plotted.

```python

Delta = 1/100
B = 1000

n = int(1 / Delta)

values = np.random.choice([-1, 1], size=(B, n))

Q6 = np.exp(np.sqrt(Delta) * values.sum(axis=1))

sns.ecdfplot(Q6)
```

Make delta smaller:

```python
Delta = 1/1000
B = 1000

n = int(1 / Delta)

values = np.random.choice([-1, 1], size=(B, n))

Q6 = np.exp(np.sqrt(Delta) * values.sum(axis=1))

sns.ecdfplot(Q6)
```

As delta gets closer to 0 we approach a lognormal distribution.
